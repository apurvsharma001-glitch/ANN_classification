```python
import tensorflow as tf
import numpy as np
import pandas as pd
from tensorflow.keras.models import load_model
import pickle
```

```python
model = load_model("model.h5")
with open("label_encoder.pkl","rb") as f:
    label_encoder = pickle.load(f)
with open("geo_encoder.pkl","rb") as f:
    geo_encoder = pickle.load(f)
with open("scaler.pkl","rb") as f:
    scaler = pickle.load(f)
```

```python
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
geo_encodedd = geo_encoder.transform([[input_data['Geography']]]).toarray()
geo_encodedd_df = pd.DataFrame(geo_encodedd,columns=geo_encoder.get_feature_names_out(["Geography"]))
geo_encodedd_df
```

```python
input_data_df = pd.DataFrame([input_data])
input_data_df
```

```python
input_data_df["Gender"]= label_encoder.transform(input_data_df["Gender"])
```

```python
input_data_df
```

```python
final_df = pd.concat((input_data_df.drop('Geography', axis=1),geo_encodedd_df), axis=1)
```

```python
final_df
```

```python
scaled_df = scaler.transform(final_df)
pred = model.predict(scaled_df)
```
